```sql
%sql
insert overwrite table ${banfield_catalog}.bfdw_bronze.dwworkday_emplegacyid
(
    wd_employee_id,
    oracle_id,
    bp_legacy_id,
    src_filename,
    dw_create_dt,
    dw_load_dt,
    dw_job_id,
    fw_createdts,
    fw_modifiedts,
    fw_filename,
    rowhash,
    record_status
)
select
    cast(wd_employee_id as string)        as wd_employee_id,
    cast(oracle_id as string)             as oracle_id,
    cast(bp_legacy_id as string)          as bp_legacy_id,
    cast(source_file as string)           as src_filename,
    cast(dw_create_dt as timestamp)       as dw_create_dt,
    cast(dw_load_dt as timestamp)         as dw_load_dt,
    cast(dw_job_id as bigint)             as dw_job_id,

    cast(ingestion_ts as timestamp)       as fw_createdts,
    current_timestamp()                   as fw_modifiedts,
    cast(source_file as string)           as fw_filename,

    sha2(
        coalesce(trim(cast(oracle_id as string)), ''), 
        512
    )                                     as rowhash,

    case
        when oracle_id is not null then 'valid'
        else 'invalid'
    end                                   as record_status

from ${banfield_catalog}.bfdw_iron.dwworkday_emplegacyid;
```
