```python
%sql
insert overwrite table ${banfield_catalog}.bfdw_bronze.tb_cmd_invc_status
( 

 dw_invc_status_id
,invc_status_cd
,invc_status_descr
,dw_create_dt
,dw_begin_eff_dt
,dw_end_eff_dt
,dw_deleted_ind
,dw_curr_row_ind
,dw_job_id
,dw_load_dt
,fw_createdts
,fw_modifiedts
,fw_filename
,rowhash
,record_status
)
select
 cast(dw_invc_status_id as bigint)            as dw_pymnt_typ_id
,cast(invc_status_cd as string)            as pymnt_typ_descr
,cast(invc_status_descr as string) as hosp_day_summary_tab_col_nam
,cast(dw_create_dt as timestamp)            as dw_create_dt
,cast(dw_begin_eff_dt as timestamp)         as dw_begin_eff_dt
,cast(dw_end_eff_dt as timestamp)           as dw_end_eff_dt
,cast(dw_deleted_ind as bigint)             as dw_deleted_ind
,cast(dw_curr_row_ind as bigint)            as dw_curr_row_ind
,cast(dw_job_id as bigint)                  as dw_job_id
,cast(dw_load_dt as timestamp)              as dw_load_dt
,cast(ingestion_ts as timestamp) as fw_createdts
,current_timestamp as fw_modifiedts
,cast(source_file as string) as fw_filename
,sha2(COALESCE(trim(dw_pymnt_typ_id),''), 512) as rowhash
,case
    when (
          try_cast(dw_pymnt_typ_id as string) is not null 
         )
  then 'valid'
  else 'invalid'
end as record_status

from ${banfield_catalog}.bfdw_iron.tb_cmd_invc_status
;
```
